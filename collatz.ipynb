```python
def collatz(n: int) -> list[int]:
    """Gera a sequência de Collatz para um número inteiro positivo n.

    Retorna a lista completa com todos os passos até atingir o valor 1.
    """
    if n <= 0:
        raise ValueError(
            "O número deve ser um inteiro estritamente positivo (n > 0)."
        )

    sequencia = [n]
    while n != 1:
        if n % 2 == 0:
            n = n // 2
        else:
            n = 3 * n + 1
        sequencia.append(n)

    print(f"Precisou de {len(sequencia) - 1} passos para atingir o 1.")
    return sequencia
```

```python
collatz(23)
```
